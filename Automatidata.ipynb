```python
# Obtener el número de identificadores únicos de los puntos de entrega
df['DOLocationID'].nunique()
```

```python

# Calcular la distancia media por trayecto para cada punto de entrega
distance_by_dropoff = df.groupby('DOLocationID')[['trip_distance']].mean()
distance_by_dropoff

# Ordena los resultados en orden descendente según la distancia media por trayecto
distance_by_dropoff = distance_by_dropoff.sort_values(by='trip_distance')
distance_by_dropoff
```

```python
# Crea un gráfico de barras con las distancias medias de los trayectos por lugar de bajada, ordenadas de menor a mayor según la distancia.
plt.figure(figsize=(14,6))
ax = sns.barplot(x=distance_by_dropoff.index, 
                 y=distance_by_dropoff['trip_distance'],
                 order=distance_by_dropoff.index)
ax.set_xticklabels([])
ax.set_xticks([])
plt.title('Distancia media por trayecto según el lugar de bajada', fontsize=16);
```

```python
# Este gráfico presenta una curva característica relacionada con la función de densidad acumulada de una distribución normal. 
# En otras palabras, indica que los puntos de caída están distribuidos de forma relativamente uniforme por el terreno. 
# Es importante saberlo, ya que en este conjunto de datos no se incluyeron coordenadas geográficas, 
# por lo que no había una forma evidente de comprobar la distribución de las ubicaciones. 
# 
# Para confirmar esta conclusión, consideremos el siguiente experimento:
# 1. Crear una muestra de coordenadas a partir de una distribución normal —en este caso, 1.500 pares de puntos de una distribución normal 
# con una media de 10 y una desviación estándar de 5
# 2. Calcula la distancia entre cada par de coordenadas 
# 3. Agrupa las coordenadas por extremo y calcula la distancia media entre ese extremo y todos los demás puntos con los que se ha emparejado
# 4. Traza la distancia media para cada extremo único
```

```python
# 1. Generar puntos aleatorios en un plano bidimensional a partir de una distribución normal
test = np.round(np.random.normal(10, 5, (3000, 2)), 1)
midway = int(len(test)/2)  # Calcular el punto medio de la matriz de coordenadas
start = test[:midway]      # Aislar la primera mitad de la matriz («puntos de recogida»)
end = test[midway:]        # Aislar la segunda mitad de la matriz («puntos de entrega»)

# 2. Calcular las distancias euclidianas entre los puntos de la primera mitad y la segunda mitad de la matriz
distances = (start - end)**2           
distances = distances.sum(axis=-1)
distances = np.sqrt(distances)

# 3. Agrupa las coordenadas por «lugar de entrega» y calcula la distancia media.
test_df = pd.DataFrame({'start': [tuple(x) for x in start.tolist()],
                   'end': [tuple(x) for x in end.tolist()],
                   'distance': distances})

data_df = (test_df.groupby('end')[['distance']]
    .mean()
    .sort_values(by='distance', ascending=True)
    .reset_index())

# Convertir la tupla a string para que el eje la interprete como etiqueta categórica
data_df['end_label'] = data_df['end'].astype(str)

# 4. Traza la distancia media entre cada punto final («lugar de bajada») y todos los puntos con los que está conectado.
plt.figure(figsize=(14,6))
ax = sns.barplot(x=data_df.index, y=data_df['distance'])
ax.set_xticklabels([])
ax.set_xticks([])
ax.set_xlabel('Punto final')
ax.set_ylabel('Distancia media a todos los demás puntos')
ax.set_title('Distancia media entre puntos tomados al azar de una distribución normal');
```

```python
# *La curva que describe este gráfico es prácticamente idéntica a la de la distancia media recorrida por cada trayecto en taxi 
# hasta cada punto de destino. Esto revela que los puntos de destino del conjunto de datos de taxis están distribuidos geográficamente 
# de forma uniforme. Sin embargo, hay que tener en cuenta que esto *no* significa que hubiera una distribución uniforme de los *trayectos* 
# hacia cada punto de destino. Analicemos esto a continuación.

# **Histograma de viajes por punto de destino**

# En primer lugar, comprueba si los identificadores de los puntos de destino están numerados de forma consecutiva. Por ejemplo, 
#¿van del 1 al 2, del 2 al 3, del 3 al 4..., o faltan algunos números (p. ej., 1, 3, 4...)? Si los números no son todos consecutivos, 
# el histograma dará la impresión de que algunos puntos tienen muy pocos viajes o ninguno, cuando en realidad no hay ninguna barra porque 
# no existe ese punto. 
# 



```
